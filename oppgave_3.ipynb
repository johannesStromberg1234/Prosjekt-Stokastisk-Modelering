# Oppgave 3: Metropolis-algoritmen

Vi definerer konstanter $..., c_{-1}, c_0, c_1, ...$ der $0 < c_i < \infty$ for alle $i \in \Z$. Vi lar $X_0 = 0$, og definerer $\{X_n\}$  for $n = 0, 1, ...$ på følgende måte:

1. Vi genererer $X_{n+1}^*$ der $X_{n+1}^* = X_n - 1$ med sannsynlighet 0.5 og $X_n + 1$ ellers.

2. Vi genererer $U \sim Unif(0, 1)$

3. Vi lar $X_{n+1} = \begin{cases} 
X_{n+1}^*, & \text{hvis } U < \min \left\{ \dfrac{c_{X_{n+1}^*}}{c_{X_n}}, 1 \right\} \\[10pt]
X_n, & \text{ellers.} 
\end{cases}$



## Oppgave 3a:

Vi ønsker å argumentere for at $\{ X_n \}$ er en diskret markovkjede. Da må vi vise:

* At $\{ X_n \}$ er en markovprosess (markovegenskapen). 
    Det vil si $P \{X_{n+1} = j \mid X_0 = i_0, \dots, X_{n-1} = i_{n-1}, X_n = i\} = P \{X_{n+1} = j \mid X_n = i\}$.
    Dette må åpenbart være tilfelle ettersom realisasjoner av U og $X_{n+1}^*$ er uavhengige.

* At tilstandsrommet er et endelig eller tellbar mengde og at indeks-settet er $T = (0, 1, ...)$.
    Fra definisjonen ser vi at indeks-settet er nettopp $(0, 1, ...)$, og at tilstandsrommet er $\Z$ altså en tallbar mengde.


Videre ønsker vi å avgjøre om kjeden er reduserbar eller ikke, og avgjøre periodisiteten. For å gjøre det tydeliggjør vi hva som egentlig skjer i hvert steg av prosessen. Konstantene $c_i$ er definert på forhånd, og vi kan assosiere dem med en tilhørende tilstand $i \in \Z$ i tilstandsrommet. Vi foreslår i hvert steg enten å flytte oss et steg opp eller et steg ned fra nåværende tilstand $X_n$ til en tilstand $X_{n+1}^* = X_{n} \pm 1$ med 50/50 sannsynlighet, altså ett steg opp eller ned. Når dette forslaget er valgt, skjer følgende:

* Hvis $\dfrac{c_{X_{n+1}^*}}{c_{X_n}} \geq 1$, så flytter vi oss garantert.

* Hvis $\dfrac{c_{X_{n+1}^*}}{c_{X_n}} < 1$, så flytter vi oss med sannsynligheten $\dfrac{c_{X_{n+1}^*}}{c_{X_n}}$



Vi observerer umiddelbart at dersom $c_{X_{n+1}^*} = 0$, så vil vi garantert bli stående. Det vil si at dersom det finnes en konstant $c_i$ som er 0, så vil alle tilstander lenger fra 0 enn $i$ i den retningen være uoppnåelige. I alle andre tilfeller vil det alltid være en viss sannsynlighet for å bevege seg begge veier. I tilfellet der det finnes en eller flere $c_i = 0$ vil markovkjeden være reduserbar.

Videre ser vi på periodisiteten: I alle tilfeller der $\dfrac{c_{X_{n+1}^*}}{c_{X_n}} < 1$, vil det være en sannsynlighet for å bli i den samme tilstanden, og tilsvarende når $\dfrac{c_{X_{n+1}^*}}{c_{X_n}} > 1$ siden vi da i neste steg vil befinne oss i tilstanden $X_{n+1}^*$ slik at dersom $X_{n+2}^* = X_n$ så vil vi ha $\dfrac{c_{X_{n+2}^*}}{c_{X_{n+1}^*}} = \dfrac{c_{X_{n}}}{c_{X_{n+1}^*}} < 1$.

I tilfellet der alle $c_i = k, k \in \R, i \in \Z$ så vil alle $\dfrac{c_{X_{n+1}^*}}{c_{X_n}} = 1$, og antall steg man minst må ta for å komme tilbake til samme tilstand blir 2. I dette tilfellet blir altså periodisiteten 2.



## Oppgave 3b: Overgangssannsynligheter

Vi utleder her et eksplisitt uttrykk for overgangssannsynlighetene $P_{ij}$ ved å slå sammen prosessen beskrevet for å generere neste steg. VI vet at mulighetene er å ta et steg opp, ned eller bli, så resterende sannsynligheter vil alltid være 0. Vi ser på tilfellene $j = i \pm 1$ og $j = i$ hver for seg.

* $j = i \pm 1$

    Siden $U \sim Unif(0, 1)$ er $P(U < \min \left\{ \dfrac{c_{X_{n+1}^*}}{c_{X_n}}, 1 \right\}) = \min \left\{ \dfrac{c_{X_{n+1}^*}}{c_{X_n}}, 1 \right\}$. Gitt j tar vi altså et steg med sannsynligheten $\min \left\{ \dfrac{c_j}{c_i}, 1 \right\}$, og siden retningen på steget velges 50/50 har vi:

    $P_{i, j} = 0.5 \cdot \min \left\{ \dfrac{c_j}{c_i}, 1 \right\}$

* $j = i$

    Sannsynligheten for å stå igjen er den resterende sannsynligheten:

    $P_{i, i} = 1 - P_{i, i+1} - P_{i, i-1} = 1 - 0.5 \cdot \min \left\{ \dfrac{c_{i+1}}{c_i}, 1 \right\} - 0.5 \cdot \min \left\{ \dfrac{c_{i-1}}{c_i}, 1 \right\}$


Dette gir til sammen
$
P_{ij} = \begin{cases} 
0.5 \cdot \min \left\{ 1, \frac{c_j}{c_i} \right\}, & j = i \pm 1 \\ 
1 - 0.5 \left( \min \left\{ 1, \frac{c_{i+1}}{c_i} \right\} + \min \left\{ 1, \frac{c_{i-1}}{c_i} \right\} \right), & j = i \\ 
0, & \text{ellers.} 
\end{cases}
$

    

## Oppgave 3c:

Vi ønsker å vise at dersom $\{ X_n \}$ er positivt rekurrent, så er andelen av tiden bruk i tilstand $i$ $\frac{c_i}{\sum_i c_i}$. Vi antar at $\{ X_n \}$ er positivt rekurrent, da har vi at $\sum_{i=-\infty}^{\infty} c_i < \infty$.

Vi har at for positivt rekurrente Markovkjeder gir den stasjonære fordelingen $\pi = (\pi)_{i \in \Z}$ andelene av tiden brukt i tilstand $i$, så vi ønsker med andre ord å vise at $\pi_i = \frac{c_i}{\sum_i c_i}$ er en stasjonær fordeling.

Kravet om normering er åpenbart oppfylt, $\sum_i \pi_i = 1$. Videre er $\pi_i \geq 0, i \in \Z$.

Da gjenstår bare å vise $\pi P = \pi$.

Siden $P_{i, j} = 0 ,  j \not \in \{i-1, i, i+1\}$ har vi for alle $i \in \Z$

$\pi_{i} = \pi_{i-1}P_{i-1, i} + \pi_{i}P_{i, i} + \pi_{i+1}P_{i+1, i} $

$\pi_{i} = \frac{\pi_{i-1}P_{i-1, i} + \pi_{i+1}P_{i+1, i}}{1-P_{i, i}}$

Vi bruker at $P_{i-1, i} = 0.5 \min \left\{ 1, \frac{c_{i}}{c_{i-1}} \right\}$ og $P_{i+1, i} = 0.5 \min \left\{ 1, \frac{c_{i}}{c_{i+1}} \right\}$ og $P_{i, i} = 1 - 0.5 \left( \min \left\{ 1, \frac{c_{i+1}}{c_i} \right\} + \min \left\{ 1, \frac{c_{i-1}}{c_i} \right\} \right)$ og $\pi_{i-1} = \frac{c_{i-1}}{\sum_i c_i}$ og $\pi_{i+1} = \frac{c_{i+1}}{\sum_i c_i}$

$\pi_{i} = \frac{\pi_{i-1}P_{i-1, i} + \pi_{i+1}P_{i+1, i}}{1-P_{i, i}} = 
\frac{c_{i-1} \min \left\{ 1, \frac{c_{i}}{c_{i-1}} \right\} + c_{i+1}\min \left\{ 1, \frac{c_{i}}{c_{i+1}} \right\} }
{\sum_i c_i \left( \min \left\{ 1, \frac{c_{i+1}}{c_i} \right\} + \min \left\{ 1, \frac{c_{i-1}}{c_i} \right\} \right)} = 
\frac{\min \left\{ c_{i-1}, c_{i} \right\} + \min \left\{ c_{i+1}, c_{i} \right\} }
{\sum_i c_i \left( \min \left\{ 1, \frac{c_{i+1}}{c_i} \right\} + \min \left\{ 1, \frac{c_{i-1}}{c_i} \right\} \right)} $

Vi ganger med $c_{i}$ oppe og nede og får:

$\pi_{i} = c_{i} \frac{\min \left\{ c_{i-1}, c_{i} \right\} + \min \left\{ c_{i+1}, c_{i} \right\} )}
{\sum_i c_i \left( \min \left\{ c_{i}, c_{i+1} \right\} + \min \left\{ c_{i}, c_{i-1} \right\} \right)} = \frac{c_{i}}{\sum_i c_i}$
